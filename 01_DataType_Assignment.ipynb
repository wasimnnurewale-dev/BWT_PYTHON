## Q1. Add two lists index-wise
Pair items at the same index; any leftover items from the longer list are appended at the end.

```python
def add_index_wise(l1, l2):
    result = [[a, b] for a, b in zip(l1, l2)]      # paired items
    shorter = min(len(l1), len(l2))
    result.extend(l1[shorter:])                    # leftovers from list 1 (if any)
    result.extend(l2[shorter:])                    # leftovers from list 2 (if any)
    return result

list1 = ["M", "na", "i", "Ra"]
list2 = ["y", "me", "s", "hul"]
print(add_index_wise(list1, list2))
print(add_index_wise([1, 2, 3, 4], ["a", "b"]))    # leftover demo
```

## Q4. Running sum of a list

```python
def running_sum(nums):
    result, total = [], 0
    for n in nums:
        total += n
        result.append(total)
    return result

print(running_sum([1, 2, 3, 4, 5, 6]))
```

## Q5. Sum of each element and all elements greater than it
The worked example in the question (`[2,4,6,10,1] -> [22,20,16,10,23]`) shows the real rule:
for every element, add **itself + every element greater than it**. (The sentence in the text lists `10` instead of `16` for the third item – that is a typo; `6+10 = 16`.)

```python
def sum_with_greater(nums):
    return [sum(y for y in nums if y >= x) for x in nums]

print(sum_with_greater([2, 4, 6, 10, 1]))
```

## Q7. Sort alphanumeric strings by product of the digits in them
* product of all digit characters in a string (no digits → product `1`)
* sorted from **highest to lowest** product (this matches the expected output); Python's sort is stable so ties keep input order.

```python
def digit_product(s):
    prod = 1
    for ch in s:
        if ch.isdigit():
            prod *= int(ch)
    return prod

data = ['1ac21', '23fg', '456', '098d', '1', 'kls']
print(sorted(data, key=digit_product, reverse=True))
```

## Q9. Shortlist students for a job role
Records are stored as tuples `(name, higher_education, primary_skill, year)`. Matching is case-insensitive so `B.Tech` equals `b.tech`.

```python
def shortlist_students():
    n = int(input("Enter No of records- "))
    records = []
    for i in range(1, n + 1):
        print(f"Enter Details of student-{i}")
        name = input("Enter Student name- ")
        edu = input("Enter Higher Education- ")
        skill = input("Enter Primary Skill- ")
        year = input("Enter Year of Graduation- ")
        records.append((name, edu, skill, year))

    print("\nEnter Job Role Requirement")
    req_skill = input("Enter Skill- ").strip().lower()
    req_edu = input("Enter Higher Education- ").strip().lower()
    req_year = input("Enter Year of Graduation- ").strip()

    matches = [r for r in records
               if r[2].lower() == req_skill and r[1].lower() == req_edu and r[3] == req_year]

    print("\nOutput")
    if matches:
        for m in matches:
            print(m)
    else:
        print("No such candidate")

run_with_inputs(shortlist_students, [
    "2",
    "Manohar", "B.Tech", "Python", "2022",
    "Ponian", "B.Sc.", "C++", "2020",
    "Python", "B.Tech", "2022",
])
```

## Q11. Count unique vowels in a string using sets
Upper-case and lower-case vowels count as different.

```python
def unique_vowels(text):
    return {ch for ch in text if ch in "aeiouAEIOU"}

Str1 = "hands-on data science mentorship progrAm with live classes at affordable fee only on CampusX"
vowels = unique_vowels(Str1)
print("Vowels found:", sorted(vowels))
print("No of unique vowels-" + str(len(vowels)))
```

## Q12. Intersection of two sets using list comprehension
Inputs are sets (so duplicates vanish automatically). Order of a set is not guaranteed, so the result order may differ from the sample – the *elements* are the same.

```python
def intersection(a, b):
    return [x for x in a if x in b]

lst1 = {15, 9, 10, 56, 23, 78, 5, 4, 9}
lst2 = {9, 4, 5, 36, 47, 26, 10, 45, 87}
print(intersection(lst1, lst2))

lst1 = {4, 9, 1, 17, 11, 26, 28, 54, 69}
lst2 = {9, 9, 74, 21, 45, 11, 63, 28, 26}
print(intersection(lst1, lst2))
```

## Q13. Convert a list of tuples into a dictionary (values as lists)
`setdefault` also handles repeated keys by collecting all their values.

```python
def tuples_to_dict(pairs):
    d = {}
    for key, value in pairs:
        d.setdefault(key, []).append(value)
    return d

print(tuples_to_dict([("akash", 10), ("gaurav", 12), ("anand", 14), ("suraj", 20), ("akhil", 25), ("ashish", 30)]))
print(tuples_to_dict([('A', 1), ('B', 2), ('C', 3)]))
```

## Q14. Function returning unique elements (order preserved)

## Q16. Perfect number check
A number is perfect if the sum of its proper divisors equals the number (equivalently, half the sum of all divisors including itself).

```python
def is_perfect(n):
    if n < 2:
        return False
    total = 1                       # 1 divides everything
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            total += i
            if i != n // i:         # add the paired divisor
                total += n // i
    return total == n

print([n for n in range(1, 10000) if is_perfect(n)])   # 6, 28, 496, 8128
```

## Q17. Concatenate any number of dictionaries
`*dicts` accepts any number of arguments; later dictionaries overwrite duplicate keys.

```python
def concat_dicts(*dicts):
    result = {}
    for d in dicts:
        result.update(d)
    return result

dic1 = {1: 10, 2: 20}
dic2 = {3: 30, 4: 40}
dic3 = {5: 50, 6: 60}
print(concat_dicts(dic1, dic2, dic3))
```

## Q18. Histogram with bin size 10
Bin label for a number `n` is `(n-1)//10*10 + 1` to `(n-1)//10*10 + 10`, so 10 → 1-10, 11 → 11-20, 50 → 41-50.

```python
def histogram(nums, bin_size=10):
    bins = {}
    for n in nums:
        low = (n - 1) // bin_size * bin_size + 1
        label = f"{low}-{low + bin_size - 1}"
        bins[label] = bins.get(label, 0) + 1
    # sort by lower bound
    return dict(sorted(bins.items(), key=lambda kv: int(kv[0].split('-')[0])))

result = histogram([13, 42, 15, 37, 22, 39, 41, 50])
print("{" + ",".join(f"{k}:{v}" for k, v in result.items()) + "}")
```
