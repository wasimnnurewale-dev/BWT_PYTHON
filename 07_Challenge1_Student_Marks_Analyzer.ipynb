# Python Challenge 1 – Student Marks Analyzer
Enter student names with marks, then view a report with:
* average marks
* highest marks and the student(s) who scored it
* lowest marks and the student(s) who scored it
* total number of students

**Bonus features included:** all data is entered first and the report is shown afterwards · report-style layout · duplicate names are rejected (case-insensitive).

Design: small, testable functions (`add_student`, `analyze`, `print_report`) + an interactive `main()`.

```python
def add_student(students, name, marks):
    """Add a student to the dict. Returns (ok, message). Rejects duplicates and bad input."""
    name = name.strip()
    if not name:
        return False, "Name cannot be empty."
    if name.lower() in {n.lower() for n in students}:
        return False, f"'{name}' already exists – duplicate names are not allowed."
    try:
        marks = int(marks)
    except (ValueError, TypeError):
        return False, "Marks must be a whole number."
    if marks < 0:
        return False, "Marks cannot be negative."
    students[name] = marks
    return True, f"Added {name} ({marks})."
```

## Analysis logic
Ties are handled: if several students share the highest (or lowest) mark, **all** of them are listed.

```python
def analyze(students):
    marks = list(students.values())
    highest, lowest = max(marks), min(marks)
    return {
        "total": len(marks),
        "average": sum(marks) / len(marks),
        "highest": highest,
        "top_students": [n for n, m in students.items() if m == highest],
        "lowest": lowest,
        "bottom_students": [n for n, m in students.items() if m == lowest],
    }
```

## Report layout

```python
def print_report(students):
    if not students:
        print("No students were entered.")
        return
    r = analyze(students)
    width = 46
    print("=" * width)
    print("STUDENT MARKS REPORT".center(width))
    print("=" * width)
    print(f"{'#':<4}{'Student':<24}{'Marks':>8}")
    print("-" * width)
    for i, (name, m) in enumerate(students.items(), 1):
        print(f"{i:<4}{name:<24}{m:>8}")
    print("-" * width)
    print(f"{'Total students':<28}{r['total']:>18}")
    print(f"{'Average marks':<28}{r['average']:>18.2f}")
    print(f"{'Highest marks':<28}{r['highest']:>18}")
    print(f"{'   scored by':<28}{', '.join(r['top_students']):>18}")
    print(f"{'Lowest marks':<28}{r['lowest']:>18}")
    print(f"{'   scored by':<28}{', '.join(r['bottom_students']):>18}")
    print("=" * width)
```

## Interactive program
Type `done` as the student name to finish entering data; the report is printed afterwards.

```python
def main():
    students = {}
    print("Enter student data (type 'done' as the name to finish).")
    while True:
        name = input("Student name: ").strip()
        if name.lower() == "done":
            break
        marks = input(f"Marks for {name}: ")
        ok, msg = add_student(students, name, marks)
        print("  ->", msg)
    print()
    print_report(students)
```

## Demo (scripted input)
Includes a duplicate (`ravi`), a non-numeric mark and a tie for highest marks.

```python
from unittest.mock import patch

def run_with_inputs(func, inputs):
    it = iter(inputs)
    def fake_input(prompt=""):
        value = next(it)
        print(f"{prompt}{value}")
        return value
    with patch("builtins.input", fake_input):
        return func()

run_with_inputs(main, [
    "Asha", "88",
    "Ravi", "72",
    "ravi", "90",         # duplicate name -> rejected
    "Meena", "abc",       # invalid marks -> rejected
    "Meena", "95",
    "Karan", "95",        # ties with Meena for highest
    "Zoya", "61",
    "done",
])
```

## Quick unit checks

```python
s = {}
add_student(s, "A", 50); add_student(s, "B", 70); add_student(s, "C", 70)
res = analyze(s)
assert res["total"] == 3
assert abs(res["average"] - 63.33) < 0.01
assert res["top_students"] == ["B", "C"] and res["bottom_students"] == ["A"]
assert add_student(s, "a", 10)[0] is False        # duplicate (case-insensitive)
print("All checks passed ✔")
```
