## Q1. Circle class – area and perimeter

```python
import math

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius

c = Circle(7)
print(f"Radius: {c.radius}")
print(f"Area: {c.area():.2f}")
print(f"Perimeter: {c.perimeter():.2f}")
```

## Q2. Person class – calculate age from date of birth

```python
from datetime import date

class Person:
    def __init__(self, name, country, date_of_birth):
        self.name = name
        self.country = country
        self.date_of_birth = date_of_birth      # a datetime.date

    def age(self, today=None):
        today = today or date.today()
        years = today.year - self.date_of_birth.year
        # subtract one if the birthday has not happened yet this year
        if (today.month, today.day) < (self.date_of_birth.month, self.date_of_birth.day):
            years -= 1
        return years

p = Person("Asha", "India", date(1998, 5, 21))
print(f"{p.name} from {p.country} is {p.age()} years old")
print("Age on 2020-01-01:", p.age(date(2020, 1, 1)))
```

## Q4. Shape base class with Circle, Triangle and Square subclasses
`Shape` is an abstract base class; each subclass overrides `area()` and `perimeter()` (polymorphism).

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self): ...

    @abstractmethod
    def perimeter(self): ...

    def describe(self):
        return f"{self.__class__.__name__}: area={self.area():.2f}, perimeter={self.perimeter():.2f}"

class CircleShape(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return math.pi * self.r ** 2
    def perimeter(self):
        return 2 * math.pi * self.r

class Triangle(Shape):
    def __init__(self, a, b, c):
        if a + b <= c or a + c <= b or b + c <= a:
            raise ValueError("Invalid triangle sides")
        self.a, self.b, self.c = a, b, c
    def area(self):                      # Heron's formula
        s = self.perimeter() / 2
        return math.sqrt(s * (s - self.a) * (s - self.b) * (s - self.c))
    def perimeter(self):
        return self.a + self.b + self.c

class Square(Shape):
    def __init__(self, side):
        self.side = side
    def area(self):
        return self.side ** 2
    def perimeter(self):
        return 4 * self.side

for shape in [CircleShape(5), Triangle(3, 4, 5), Square(6)]:
    print(shape.describe())
```

## Q5. Binary Search Tree – insert and search
Also includes an in-order traversal, which prints the values in sorted order.

```python
class BSTNode:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, key):
        if self.root is None:
            self.root = BSTNode(key)
            return
        node = self.root
        while True:
            if key < node.key:
                if node.left is None:
                    node.left = BSTNode(key); return
                node = node.left
            elif key > node.key:
                if node.right is None:
                    node.right = BSTNode(key); return
                node = node.right
            else:
                return                      # ignore duplicates

    def search(self, key):
        node = self.root
        while node:
            if key == node.key:
                return True
            node = node.left if key < node.key else node.right
        return False

    def inorder(self):
        result = []
        def walk(n):
            if n:
                walk(n.left); result.append(n.key); walk(n.right)
        walk(self.root)
        return result

bst = BinarySearchTree()
for v in [50, 30, 70, 20, 40, 60, 80]:
    bst.insert(v)
print("In-order:", bst.inorder())
print("Search 60:", bst.search(60))
print("Search 65:", bst.search(65))
```

## Q6. Stack – push and pop

```python
class Stack:
    def __init__(self):
        self._items = []

    def push(self, item):
        self._items.append(item)

    def pop(self):
        if self.is_empty():
            raise IndexError("Stack is empty")
        return self._items.pop()

    def peek(self):
        return self._items[-1] if self._items else None

    def is_empty(self):
        return len(self._items) == 0

s = Stack()
for x in [10, 20, 30]:
    s.push(x)
print("Popped:", s.pop())
print("Popped:", s.pop())
print("Top now:", s.peek())
```

## Q7. Linked list – display, insert and delete nodes

```python
class ListNode:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def insert_at_end(self, data):
        new = ListNode(data)
        if not self.head:
            self.head = new
            return
        cur = self.head
        while cur.next:
            cur = cur.next
        cur.next = new

    def insert_at_beginning(self, data):
        new = ListNode(data)
        new.next = self.head
        self.head = new

    def delete(self, data):
        """Delete first node holding `data`. Returns True if deleted."""
        cur, prev = self.head, None
        while cur:
            if cur.data == data:
                if prev:
                    prev.next = cur.next
                else:
                    self.head = cur.next
                return True
            prev, cur = cur, cur.next
        return False

    def display(self):
        items, cur = [], self.head
        while cur:
            items.append(str(cur.data))
            cur = cur.next
        print(" -> ".join(items) + " -> None" if items else "List is empty")

ll = LinkedList()
for x in [1, 2, 3, 4]:
    ll.insert_at_end(x)
ll.insert_at_beginning(0)
ll.display()
ll.delete(3)
ll.display()
print("Delete 99:", ll.delete(99))
```

## Q8. Shopping cart – add, remove, total

```python
class ShoppingCart:
    def __init__(self):
        self.items = {}                 # name -> (price, quantity)

    def add_item(self, name, price, quantity=1):
        if name in self.items:
            old_price, old_qty = self.items[name]
            self.items[name] = (price, old_qty + quantity)
        else:
            self.items[name] = (price, quantity)

    def remove_item(self, name, quantity=None):
        """Remove the item completely, or only `quantity` units of it."""
        if name not in self.items:
            print(f"{name} is not in the cart")
            return
        price, qty = self.items[name]
        if quantity is None or quantity >= qty:
            del self.items[name]
        else:
            self.items[name] = (price, qty - quantity)

    def total_price(self):
        return sum(price * qty for price, qty in self.items.values())

    def show(self):
        for name, (price, qty) in self.items.items():
            print(f"{name:10} {qty} x {price:7.2f} = {price * qty:8.2f}")
        print(f"{'TOTAL':10} {'':14} {self.total_price():8.2f}")

cart = ShoppingCart()
cart.add_item("Laptop", 55000)
cart.add_item("Mouse", 500, 2)
cart.add_item("Keyboard", 1200)
cart.remove_item("Mouse", 1)
cart.show()
cart.remove_item("Laptop")
print("After removing laptop, total =", cart.total_price())
```

## Q10. Queue – enqueue and dequeue

```python
from collections import deque

class Queue:
    def __init__(self):
        self._items = deque()

    def enqueue(self, item):
        self._items.append(item)

    def dequeue(self):
        if not self._items:
            raise IndexError("Queue is empty")
        return self._items.popleft()

    def size(self):
        return len(self._items)

    def display(self):
        print("Front ->", list(self._items), "<- Rear")

q = Queue()
for x in ["Ann", "Bob", "Cara"]:
    q.enqueue(x)
q.display()
print("Dequeued:", q.dequeue())
q.display()
```

## Q11. Bank and BankAccount
`BankAccount` follows the spec: `accountNumber`, `name`, `balance`, `Deposit()`, `Withdrawal()`, `bankFees()` (5 % of balance) and `display()`. A `Bank` class manages many accounts and transactions.

```python
class BankAccount:
    def __init__(self, accountNumber, name, balance):
        self.accountNumber = accountNumber
        self.name = name
        self.balance = balance

    def Deposit(self, amount):
        if amount <= 0:
            print("Deposit amount must be positive")
            return
        self.balance += amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def Withdrawal(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive")
        elif amount > self.balance:
            print("Insufficient funds")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")

    def bankFees(self):
        fee = self.balance * 0.05
        self.balance -= fee
        print(f"Bank fees of {fee:.2f} (5%) applied. New balance: {self.balance:.2f}")

    def display(self):
        print(f"Account Number: {self.accountNumber} | Owner: {self.name} | Balance: {self.balance:.2f}")


class Bank:
    def __init__(self, name):
        self.name = name
        self.accounts = {}

    def open_account(self, accountNumber, name, balance=0):
        if accountNumber in self.accounts:
            print("Account number already exists")
            return None
        acc = BankAccount(accountNumber, name, balance)
        self.accounts[accountNumber] = acc
        return acc

    def get_account(self, accountNumber):
        return self.accounts.get(accountNumber)

    def transfer(self, from_no, to_no, amount):
        src, dst = self.get_account(from_no), self.get_account(to_no)
        if not src or not dst:
            print("Invalid account number")
        elif amount > src.balance:
            print("Transfer failed: insufficient funds")
        else:
            src.balance -= amount
            dst.balance += amount
            print(f"Transferred {amount} from {from_no} to {to_no}")

    def close_account(self, accountNumber):
        return self.accounts.pop(accountNumber, None) is not None

    def display_all(self):
        print(f"--- {self.name} ---")
        for acc in self.accounts.values():
            acc.display()

# BankAccount demo
acc = BankAccount(1001, "Ravi", 10000)
acc.Deposit(5000)
acc.Withdrawal(2000)
acc.Withdrawal(50000)
acc.bankFees()
acc.display()

# Bank demo
print()
bank = Bank("National Bank")
bank.open_account(1, "Alice", 5000)
bank.open_account(2, "Bob", 1000)
bank.transfer(1, 2, 1500)
bank.display_all()
```

## Q13. TechWorld instructor course allocation
All instance variables are **private**; setters/getters are public. Eligibility:
* experience > 3 years → average feedback ≥ 4.5
* experience ≤ 3 years → average feedback ≥ 4
* a course can be allocated only if the instructor is eligible **and** has the required skill.

```python
class Instructor:
    def __init__(self):
        self.__name = None
        self.__technology_skill = []
        self.__experience = 0
        self.__avg_feedback = 0.0

    # setters
    def set_name(self, name): self.__name = name
    def set_technology_skill(self, skills): self.__technology_skill = list(skills)
    def set_experience(self, exp): self.__experience = exp
    def set_avg_feedback(self, fb): self.__avg_feedback = fb

    # getters
    def get_name(self): return self.__name
    def get_technology_skill(self): return self.__technology_skill
    def get_experience(self): return self.__experience
    def get_avg_feedback(self): return self.__avg_feedback

    def check_eligibility(self):
        if self.__experience > 3:
            return self.__avg_feedback >= 4.5
        return self.__avg_feedback >= 4

    def allocate_course(self, technology):
        return self.check_eligibility() and technology in self.__technology_skill


def make_instructor(name, skills, exp, fb):
    i = Instructor()
    i.set_name(name); i.set_technology_skill(skills)
    i.set_experience(exp); i.set_avg_feedback(fb)
    return i

team = [
    make_instructor("Priya",  ["Python", "Java"], 5, 4.7),
    make_instructor("Rahul",  ["C++"],            2, 4.2),
    make_instructor("Meena",  ["Python"],         6, 4.3),    # too low for >3 yrs
    make_instructor("Karan",  ["Java", "SQL"],    3, 3.9),    # too low for <=3 yrs
]

for t in team:
    print(f"{t.get_name():6} eligible={t.check_eligibility()!s:5} "
          f"Python={t.allocate_course('Python')!s:5} Java={t.allocate_course('Java')!s:5}")
```

## Q14. Years, months and days left until expiry (datetime inside a class)
Dates are entered as `YYYY-MM-DD`. The remaining time is calculated from a reference date (today by default; we pass a fixed date in the demo so the output never changes).

```python
from datetime import datetime, date
import calendar

class Product:
    def __init__(self, name, manufacturing_date, expiry_date):
        self.name = name
        self.mfg = datetime.strptime(manufacturing_date, "%Y-%m-%d").date()
        self.exp = datetime.strptime(expiry_date, "%Y-%m-%d").date()
        if self.exp <= self.mfg:
            raise ValueError("Expiry date must be after manufacturing date")

    def time_left(self, today=None):
        today = today or date.today()
        if today >= self.exp:
            return None
        years = self.exp.year - today.year
        months = self.exp.month - today.month
        days = self.exp.day - today.day
        if days < 0:                                    # borrow days from previous month
            months -= 1
            prev_month = self.exp.month - 1 or 12
            prev_year = self.exp.year if self.exp.month > 1 else self.exp.year - 1
            days += calendar.monthrange(prev_year, prev_month)[1]
        if months < 0:
            years -= 1
            months += 12
        return years, months, days

    def report(self, today=None):
        left = self.time_left(today)
        if left is None:
            print(f"{self.name} has already expired!")
        else:
            y, m, d = left
            print(f"{self.name}: {y} year(s), {m} month(s), {d} day(s) left until expiry")

milk = Product("Milk Powder", "2025-01-15", "2027-03-10")
milk.report(today=date(2026, 10, 10))
Product("Yogurt", "2026-09-01", "2026-10-05").report(today=date(2026, 10, 10))

# Interactive version (uncomment to use):
# Product(input("Name: "), input("Manufacturing date (YYYY-MM-DD): "), input("Expiry date (YYYY-MM-DD): ")).report()
```

## Q15. University admission

## Q16. Ice-cream Scoops and Bowl shop
* `Scoop`: public `flavor`, private `__price` (getter + setter) and a `sold()` that reports total scoops sold.
* `Bowl`: private `__scoop_list`, `add_scoops(*scoops)`, `display()` and a `sold()` that reports bowls sold.
* A scoop counts as sold when it is put in a bowl, and a bowl counts as sold when it is created (one customer order).

```python
class Scoop:
    _scoops_sold = 0

    def __init__(self, flavor, price):
        self.flavor = flavor
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, price):
        if price < 0:
            raise ValueError("Price cannot be negative")
        self.__price = price

    @classmethod
    def sold(cls):
        print(f"Total scoops sold: {cls._scoops_sold}")


class Bowl:
    _bowls_sold = 0

    def __init__(self):
        self.__scoop_list = []
        Bowl._bowls_sold += 1

    def add_scoops(self, *scoops):
        for s in scoops:
            self.__scoop_list.append(s)
            Scoop._scoops_sold += 1

    def display(self):
        total = 0
        for s in self.__scoop_list:
            print(f"Flavour: {s.flavor:12} Price: {s.get_price()}")
            total += s.get_price()
        print(f"Total price of the bowl: {total}")

    @classmethod
    def sold(cls):
        print(f"Total bowls sold: {cls._bowls_sold}")


vanilla, choco, mango = Scoop("Vanilla", 40), Scoop("Chocolate", 50), Scoop("Mango", 45)
bowl1 = Bowl()
bowl1.add_scoops(vanilla, choco, mango)
bowl1.display()

choco.set_price(55)                     # price change through the setter
bowl2 = Bowl()
bowl2.add_scoops(Scoop("Strawberry", 48), choco)
print()
bowl2.display()
print()
Scoop.sold()
Bowl.sold()
```

## Q17. Vehicle and Bus (method overriding)
Default fare = seating capacity × 100. `Bus` overrides `fare()` and adds a 10 % maintenance charge, so a 50-seat bus costs 5500.

```python
class Vehicle:
    def __init__(self, name, mileage, capacity):
        self.name = name
        self.mileage = mileage
        self.capacity = capacity

    def fare(self):
        return self.capacity * 100


class Bus(Vehicle):
    def fare(self):
        amount = super().fare()
        return amount + amount * 0.10


School_bus = Bus("School Volvo", 12, 50)
print("Total Bus fare is:", School_bus.fare())      # 5500.0

car = Vehicle("Audi Q5", 18, 5)
print("Total Vehicle fare is:", car.fare())         # 500
```

## Q18. Employee class
Overtime rule: hours worked above 50 are paid at `salary / 50` per hour.

```python
class Employee:
    def __init__(self, emp_name, emp_id, emp_salary, emp_department):
        self.emp_name = emp_name
        self.emp_id = emp_id
        self.emp_salary = emp_salary
        self.emp_department = emp_department

    def calculate_emp_salary(self, salary, hours_worked):
        total = salary
        if hours_worked > 50:
            overtime = hours_worked - 50
            total += overtime * (salary / 50)
        return total

    def assign_department(self, new_department):
        self.emp_department = new_department

    def print_employee_details(self):
        print(f"Name: {self.emp_name}")
        print(f"ID: {self.emp_id}")
        print(f"Salary: {self.emp_salary}")
        print(f"Department: {self.emp_department}")
        print("-" * 28)


emp1 = Employee("ADAMS", "E7876", 50000, "ACCOUNTING")
emp2 = Employee("JONES", "E7499", 45000, "RESEARCH")
emp3 = Employee("MARTIN", "E7900", 50000, "SALES")
emp4 = Employee("SMITH", "E7698", 55000, "OPERATIONS")

for e in (emp1, emp2, emp3, emp4):
    e.print_employee_details()

emp1.assign_department("OPERATIONS")
print("After department change:")
emp1.print_employee_details()

print("Salary for 55 hours (Adams):", emp1.calculate_emp_salary(emp1.emp_salary, 55))
print("Salary for 45 hours (Jones):", emp2.calculate_emp_salary(emp2.emp_salary, 45))
```

## Q19. Restaurant class (dictionaries and lists)
* `menu_items` – dict `{item: price}`
* `book_table` – list of reservations
* `customer_orders` – list of orders

```python
class Restaurant:
    def __init__(self):
        self.menu_items = {}
        self.book_table = []
        self.customer_orders = []

    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price

    def book_tables(self, table_number, customer_name):
        if any(b["table"] == table_number for b in self.book_table):
            print(f"Table {table_number} is already booked")
            return
        self.book_table.append({"table": table_number, "customer": customer_name})

    def customer_order(self, table_number, *items):
        invalid = [i for i in items if i not in self.menu_items]
        if invalid:
            print("Not on menu:", invalid)
            return
        total = sum(self.menu_items[i] for i in items)
        self.customer_orders.append({"table": table_number, "items": list(items), "total": total})

    def print_menu(self):
        print("MENU")
        for item, price in self.menu_items.items():
            print(f"  {item:15} Rs.{price}")

    def print_table_reservations(self):
        print("TABLE RESERVATIONS")
        for b in self.book_table:
            print(f"  Table {b['table']} -> {b['customer']}")

    def print_customer_orders(self):
        print("CUSTOMER ORDERS")
        for o in self.customer_orders:
            print(f"  Table {o['table']}: {', '.join(o['items'])} (Total Rs.{o['total']})")


r = Restaurant()
# 1. add items to the menu
r.add_item_to_menu("Paneer Tikka", 250)
r.add_item_to_menu("Butter Naan", 40)
r.add_item_to_menu("Dal Makhani", 180)
r.add_item_to_menu("Gulab Jamun", 80)
# 2. table reservations
r.book_tables(1, "Sharma")
r.book_tables(4, "Patel")
r.book_tables(1, "Gupta")                  # duplicate booking rejected
# 3. customer orders
r.customer_order(1, "Paneer Tikka", "Butter Naan", "Butter Naan")
r.customer_order(4, "Dal Makhani", "Gulab Jamun")
r.customer_order(4, "Pizza")               # not on menu
# 4-6. print everything
print()
r.print_menu()
r.print_table_reservations()
r.print_customer_orders()
```

## Q20. BankAccount with deposit, withdraw and check_balance

```python
from datetime import date

class SimpleBankAccount:
    def __init__(self, account_number, customer_name, balance=0, date_of_opening=None):
        self.account_number = account_number
        self.customer_name = customer_name
        self.balance = balance
        self.date_of_opening = date_of_opening or date.today()

    def deposit(self, amount):
        if amount <= 0:
            print("Invalid deposit amount")
            return
        self.balance += amount
        print(f"Deposited {amount}")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")
        elif amount > self.balance:
            print("Insufficient balance")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}")

    def check_balance(self):
        print(f"{self.customer_name} (A/c {self.account_number}, opened {self.date_of_opening}) "
              f"- Balance: {self.balance}")

a = SimpleBankAccount("AC1001", "Neha", 2000, date(2023, 4, 1))
a.check_balance()
a.deposit(3000)
a.withdraw(1200)
a.withdraw(9999)
a.check_balance()
```

## Q21. Inventory class
Items are kept in a dictionary keyed by `item_id`; each record has `item_name`, `stock_count` and `price`.

```python
class Inventory:
    def __init__(self):
        self.items = {}

    def add_item(self, item_id, item_name, stock_count, price):
        if item_id in self.items:
            print(f"Item {item_id} already exists – use update_item")
            return
        self.items[item_id] = {"item_name": item_name, "stock_count": stock_count, "price": price}

    def update_item(self, item_id, stock_count=None, price=None, item_name=None):
        if item_id not in self.items:
            print(f"Item {item_id} not found")
            return
        if stock_count is not None: self.items[item_id]["stock_count"] = stock_count
        if price is not None:       self.items[item_id]["price"] = price
        if item_name is not None:   self.items[item_id]["item_name"] = item_name

    def check_item_details(self, item_id):
        item = self.items.get(item_id)
        if not item:
            print(f"Item {item_id} not found")
            return
        print(f"ID: {item_id} | Name: {item['item_name']} | Stock: {item['stock_count']} | Price: {item['price']}")

inv = Inventory()
inv.add_item("I001", "Notebook", 100, 45)
inv.add_item("I002", "Pen", 250, 10)
inv.add_item("I001", "Duplicate", 1, 1)
inv.check_item_details("I001")
inv.update_item("I001", stock_count=80, price=50)
inv.check_item_details("I001")
inv.check_item_details("I999")
```
