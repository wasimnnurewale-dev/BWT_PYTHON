# Python File I/O Assignment – Solutions
15 programs on reading, writing, appending and analysing text files.

We first create a sample file `sample.txt` so every program has something to work on. Replace the file name with your own file whenever you like.

```python
import os

sample_text = """Python is a powerful programming language.
It is easy to learn and widely used in data science.
File handling lets programs read and write data on disk.
Good programmers always close their files.
Python makes file handling simple and readable.
Practice makes perfect, so keep coding every day.
"""

with open("sample.txt", "w") as f:
    f.write(sample_text)

print("sample.txt created")
```

## Q1. Read an entire text file

```python
def read_entire_file(path):
    with open(path, "r") as f:
        return f.read()

print(read_entire_file("sample.txt"))
```

## Q2. Read first *n* lines of a file

```python
def read_first_n_lines(path, n):
    lines = []
    with open(path, "r") as f:
        for i, line in enumerate(f):
            if i >= n:
                break
            lines.append(line.rstrip("\n"))
    return lines

for line in read_first_n_lines("sample.txt", 3):
    print(line)
```

## Q3. Append text to a file and display it
Mode `"a"` adds to the end without erasing existing content.

```python
def append_and_display(path, text):
    with open(path, "a") as f:
        f.write(text + "\n")
    with open(path, "r") as f:
        print(f.read())

append_and_display("sample.txt", "This line was appended later.")
```

## Q4. Read last *n* lines of a file
`collections.deque(maxlen=n)` keeps only the last *n* lines in memory, so it works even for huge files.

```python
from collections import deque

def read_last_n_lines(path, n):
    with open(path, "r") as f:
        return [line.rstrip("\n") for line in deque(f, maxlen=n)]

for line in read_last_n_lines("sample.txt", 2):
    print(line)
```

## Q5. Read a file line by line and store it in a list

```python
def file_to_list(path):
    lines = []
    with open(path, "r") as f:
        for line in f:
            lines.append(line.rstrip("\n"))
    return lines

lines_list = file_to_list("sample.txt")
print(lines_list)
```

## Q6. Read a file line by line and store it in a variable
Each line is concatenated into one string variable.

```python
def file_to_variable(path):
    content = ""
    with open(path, "r") as f:
        for line in f:
            content += line
    return content

text_var = file_to_variable("sample.txt")
print(type(text_var))
print(text_var)
```

## Q7. Find the longest word(s)
Punctuation is stripped; if several words tie for the maximum length, all are returned.

```python
import string

def longest_words(path):
    with open(path, "r") as f:
        words = [w.strip(string.punctuation) for w in f.read().split()]
    words = [w for w in words if w]
    max_len = max(len(w) for w in words)
    return sorted({w for w in words if len(w) == max_len})

print(longest_words("sample.txt"))
```

## Q8. Count the number of lines

```python
def count_lines(path):
    with open(path, "r") as f:
        return sum(1 for _ in f)

print("Number of lines:", count_lines("sample.txt"))
```

## Q9. Frequency of words in a file
Words are lower-cased and stripped of punctuation so "Python" and "python." count as the same word.

```python
from collections import Counter

def word_frequency(path):
    with open(path, "r") as f:
        words = [w.strip(string.punctuation).lower() for w in f.read().split()]
    return Counter(w for w in words if w)

freq = word_frequency("sample.txt")
for word, count in freq.most_common(8):
    print(f"{word:<12}{count}")
```

## Q10. Get the file size

```python
def get_file_size(path):
    return os.path.getsize(path)      # size in bytes

print(f"sample.txt is {get_file_size('sample.txt')} bytes")
```

## Q11. Write a list to a file

```python
def write_list_to_file(path, items):
    with open(path, "w") as f:
        for item in items:
            f.write(f"{item}\n")

fruits = ["apple", "banana", "cherry", "mango"]
write_list_to_file("fruits.txt", fruits)
print(read_entire_file("fruits.txt"))
```

## Q12. Copy the contents of one file to another

```python
def copy_file(src, dst):
    with open(src, "r") as fin, open(dst, "w") as fout:
        for line in fin:
            fout.write(line)

copy_file("sample.txt", "sample_copy.txt")
print("Copy identical to original:", read_entire_file("sample.txt") == read_entire_file("sample_copy.txt"))
```

## Q13. Read a random line from a file

```python
import random

def random_line(path):
    with open(path, "r") as f:
        lines = f.read().splitlines()
    return random.choice(lines)

print(random_line("sample.txt"))
```

## Q14. Check whether a file is closed
A file object has a `.closed` attribute – `False` while open, `True` after `close()` (or after leaving a `with` block).

```python
f = open("sample.txt", "r")
print("Inside open():", f.closed)
f.close()
print("After close():", f.closed)

with open("sample.txt", "r") as g:
    print("Inside with:", g.closed)
print("After with block:", g.closed)
```

## Q15. Count the number of words in a text file

```python
def count_words(path):
    with open(path, "r") as f:
        return len(f.read().split())

print("Number of words:", count_words("sample.txt"))
```
